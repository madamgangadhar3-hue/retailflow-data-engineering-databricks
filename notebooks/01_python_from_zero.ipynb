## 01-01: Give a value a name

**Core concept Vibe:** A variable holds a value we can reuse.

**Build Buddy action:** Create variables for the sale ID and its quantity.

**Expected checkpoint:** S0003 and quantity 3.


```python
# Build Buddy step 01-01
# A variable holds a value we can reuse.
sale_id = 'S0003'
quantity = 3
print('Sale ID:', sale_id)
print('Quantity:', quantity)
assert sale_id == 'S0003' and quantity == 3
print('CHECKPOINT 01-01 PASS: S0003 and quantity 3')
```

## 01-02: Represent money and a percentage

**Core concept Vibe:** Text, whole numbers, and decimal numbers have different uses.

**Build Buddy action:** Represent the unit price and discount. Decimal preserves exact money values.

**Expected checkpoint:** INR 150.00 per Tea 250 g pack; 10 means 10 percent.


```python
# Build Buddy step 01-02
# Text, whole numbers, and decimals serve different purposes; Decimal keeps money exact.
from decimal import Decimal
unit_price = Decimal('150.00')  # INR per Tea 250 g pack
discount_pct = 10               # whole-number percentage points
print('Unit price:', unit_price)
print('Discount percent:', discount_pct)
assert unit_price == Decimal('150.00') and discount_pct == 10
print('CHECKPOINT 01-02 PASS: INR 150.00 per Tea 250 g pack; 10 means 10 percent')
```

## 01-03: Multiply quantity by price

**Core concept Vibe:** Multiplication calculates the amount before discounts.

**Build Buddy action:** Calculate gross sales.

**Expected checkpoint:** Gross sales = INR 450.00.


```python
# Build Buddy step 01-03
# Multiplication calculates the amount before discounts.
gross_sales = quantity * unit_price
print('Gross sales:', gross_sales)
assert gross_sales == Decimal('450.00')
print('CHECKPOINT 01-03 PASS: gross sales = INR 450.00')
```

## 01-04: Calculate the discount

**Core concept Vibe:** A percentage means a number out of 100.

**Build Buddy action:** Convert 10 percentage points to a fraction and calculate the discount amount.

**Expected checkpoint:** Discount fraction = 0.1; discount amount = INR 45.00.


```python
# Build Buddy step 01-04
# A percentage is a number out of 100.
discount_fraction = discount_pct / Decimal(100)
discount_amount = (gross_sales * discount_fraction).quantize(Decimal('0.01'))
print('Discount fraction:', discount_fraction)
print('Discount amount:', discount_amount)
assert discount_fraction == Decimal('0.1') and discount_amount == Decimal('45.00')
print('CHECKPOINT 01-04 PASS: discount fraction = 0.1; discount amount = INR 45.00')
```

## 01-05: Subtract the discount

**Core concept Vibe:** Net sales is sales after discounts and before returns.

**Build Buddy action:** Calculate and display net sales.

**Expected checkpoint:** S0003 version 1 net sales = INR 405.00.


```python
# Build Buddy step 01-05
# Net sales is after discounts and before returns.
net_sales = gross_sales - discount_amount
print('Net sales for', sale_id, '(version 1):', net_sales)
assert net_sales == Decimal('405.00')
print('CHECKPOINT 01-05 PASS: S0003 version 1 net sales = INR 405.00')
```

## 01-06: Group related values

**Core concept Vibe:** A dictionary gives names to several values in one record.

**Build Buddy action:** Create a dictionary describing this sale calculation.

**Expected checkpoint:** The dictionary holds the same values; no new sale is invented.


```python
# Build Buddy step 01-06
# A dictionary gives names to several values in one record.
sale_record = {
    'sale_id': sale_id,
    'quantity': quantity,
    'unit_price': unit_price,
    'discount_pct': discount_pct,
    'gross_sales': gross_sales,
    'discount_amount': discount_amount,
    'net_sales': net_sales,
}
print('Sale record:', sale_record)
assert sale_record['sale_id'] == sale_id and sale_record['quantity'] == quantity
assert sale_record['unit_price'] == unit_price and sale_record['discount_pct'] == discount_pct
assert sale_record['gross_sales'] == gross_sales and sale_record['discount_amount'] == discount_amount
assert sale_record['net_sales'] == net_sales
print('CHECKPOINT 01-06 PASS: the dictionary holds the same values; no new sale is invented')
```

## 01-07: Verify against the real file

**Core concept Vibe:** An answer should be traceable to its source record.

**Build Buddy action:** Read S0003 from the initial CSV and check the values used above.

**Expected checkpoint:** The manually entered values exactly match the released CSV.


```python
# Build Buddy step 01-07
# Trace the manual answer back to the released CSV record.
import csv
from pathlib import Path
with (Path(DATASET_ROOT) / 'data' / 'sales_batch_01.csv').open(encoding='utf-8', newline='') as stream:
    source_rows = list(csv.DictReader(stream))
s0003 = next(row for row in source_rows if row['sale_id'] == 'S0003' and row['record_version'] == '1')
print('S0003 version 1 source record:', s0003)
assert int(s0003['quantity']) == quantity
assert Decimal(s0003['unit_price']) == unit_price
assert Decimal(s0003['discount_pct']) == discount_pct
source_gross = int(s0003['quantity']) * Decimal(s0003['unit_price'])
source_net = source_gross - (source_gross * Decimal(s0003['discount_pct']) / Decimal(100)).quantize(Decimal('0.01'))
print('Net sales recomputed from the CSV:', source_net)
assert source_net == net_sales == Decimal('405.00')
print('CHECKPOINT 01-07 PASS: the manually entered values exactly match the released CSV')
```

What is a variable? Why divide discount_pct by 100? Why are net sales different from profit? Explain why this notebook reports version 1 and does not yet apply the Day 2 correction.

### My explanation

**What is a variable?** A variable is a name bound to a value so the value can be reused and traced. `sale_id = 'S0003'` and `quantity = 3` carry the sale's facts into every later step, and `net_sales` names the result of the arithmetic instead of leaving a bare number.

**Why divide discount_pct by 100?** `discount_pct` stores percentage points, so 10 means 10 per hundred. Dividing by `Decimal(100)` converts it to a fraction (0.1) before multiplying, which turns INR 450.00 gross sales into an INR 45.00 discount. Multiplying by 10 directly would give a meaningless 4,500.

**Why are net sales different from profit?** Net sales is the money received after discounts but before returns — INR 405.00 for S0003 version 1. Profit also subtracts costs such as purchase prices, salaries and rent, which this dataset does not contain at all. Net sales measures revenue; profit measures what is left after costs.

**Why version 1 and no Day 2 correction yet?** This notebook works from `sales_batch_01.csv` as released on Day 1, where S0003 exists only with `record_version = 1` (3 packs at INR 150.00 with a 10% discount). Checkpoint 01-07 proved the manual values match that CSV exactly. The Day 2 correction — version 2 with quantity 4 — arrives with the incremental batch and is applied by the version-guarded MERGE in notebook 05, so recomputing it here would mix course stages.
